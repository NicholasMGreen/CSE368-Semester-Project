```python
import requests

response = requests.post(
    "http://127.0.0.1:11434/api/generate",
    json={
        "model": "llama3.2:3b",
        "prompt": "What are the responsibilities of an entry level AI engineer",
        "stream": False
    }
)
print(response.json()["response"])
```

```python
job_description = """
Software Engineer - Entry Level
We are looking for an entry-level Software Engineer
to join our engineering team.
Responsibilities:
- Develop applications using Python.
- Build and maintain REST APIs.
- Work with SQL databases.
- Collaborate with other developers.
- Write unit tests and debug applications.
Requirements:
- Bachelor's degree in Computer Science
or a related field.
- Knowledge of Python and SQL.
- Understanding of data structures and algorithms.
- Familiarity with Git.
Preferred:
- Experience with Flask or Django.
- Knowledge of Docker and AWS.
- Familiarity with machine learning.
"""
```

```python
#create a prompt dinamically

def create_job_analysis_prompt(job_description):
  prompt = f"""
  Analyze the job posting.
  JOB DESCRIPTION:
  {job_description}
  Perform the following tasks:

    1. Identify the job title and seniority level.
    2. List the required technical skills.
    3. List the preferred technical skills.
    4. Identify the education and experience requirements.
    5. Summarize the main responsibilities.
    6. Generate five technical interview questions.

    Do not invent requirements that are not in the
    job description.

    Return your answer in clearly labeled sections.
    """
  return prompt
```

```python
response = requests.post(
    "http://127.0.0.1:11434/api/generate",
    json={
        "model": "llama3.2:3b",
        "prompt": create_job_analysis_prompt(job_description),
        "stream": False
    }
)
print(response.json()["response"])
```
